# Initial Data Exploration

```python
import pandas as pd

df_category = pd.read_csv(
    "../data/raw/product_category_name_translation.csv"
)

display(df_category.head())
```

```python
print("Shape:", df_category.shape)

print("\nColumn Information:")
df_category.info()

print("\nMissing Values:")
display(df_category.isnull().sum())

print("\nDuplicate Rows:", df_category.duplicated().sum())

print("\nStatistical Summary:")
display(df_category.describe(include="all"))
```

```python
df_products = pd.read_csv(
    "../data/raw/olist_products_dataset.csv"
)
```

```python
# Check duplicate category names
print("Duplicate Portuguese categories:",
      df_category["product_category_name"].duplicated().sum())

print("Duplicate English categories:",
      df_category["product_category_name_english"].duplicated().sum())

# Check mapping uniqueness
print("Unique mappings:",
      df_category.drop_duplicates().shape[0])

# Compare with products dataset
product_categories = set(df_products["product_category_name"].dropna())
translated_categories = set(df_category["product_category_name"])

missing_translations = product_categories - translated_categories

print("Product categories:", len(product_categories))
print("Translated categories:", len(translated_categories))
print("Categories without translation:", len(missing_translations))

if missing_translations:
    print(sorted(missing_translations))
```

```python
df_category_clean = df_category.copy()

# Preserve missing translations as null
df_category_clean["product_category_name"] = (
    df_category_clean["product_category_name"].astype("string")
)

df_category_clean["product_category_name_english"] = (
    df_category_clean["product_category_name_english"].astype("string")
)

# Final validation
print("Shape:", df_category_clean.shape)
print("Missing values:")
display(df_category_clean.isnull().sum())

print("Duplicate rows:", df_category_clean.duplicated().sum())
```

```python
df_category_clean.to_csv(
    "../data/processed/category_translation_clean.csv",
    index=False
)
```
