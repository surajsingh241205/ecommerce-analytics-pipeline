# Initial Data Exploration

```python
import pandas as pd
import numpy as np

df_sellers = pd.read_csv(
    "../data/raw/olist_sellers_dataset.csv",
    dtype={
        "seller_id": "string",
        "seller_zip_code_prefix": "string"
    }
)

df_sellers.head()
```

```python
print("Shape:", df_sellers.shape)

print("\nColumn Information:")
df_sellers.info()

print("\nMissing Values:")
display(df_sellers.isnull().sum())

print("\nDuplicate Rows:", df_sellers.duplicated().sum())

print("\nStatistical Summary:")
display(df_sellers.describe(include="all"))
```

# Seller distribution by state

```python
# Seller count by state
state_counts = (
    df_sellers["seller_state"]
    .value_counts()
    .reset_index()
)

state_counts.columns = ["State", "Seller_Count"]

state_counts["Percentage"] = (
    state_counts["Seller_Count"] / len(df_sellers) * 100
).round(2)

display(state_counts)
```

```python
display(df_sellers["seller_city"].value_counts().head(10))
```

# ZIP Code Distribution

```python
zip_distribution = (
    df_sellers["seller_zip_code_prefix"]
    .value_counts()
    .head(10)
    .reset_index()
)

zip_distribution.columns = ["ZIP_Code_Prefix", "Seller_Count"]

display(zip_distribution)
```

```python
# ZIP code validation
print("ZIP codes not 5 digits:",
      (~df_sellers["seller_zip_code_prefix"].str.fullmatch(r"\d{5}")).sum())

# City formatting
print("Cities not lowercase:",
      (df_sellers["seller_city"] != df_sellers["seller_city"].str.lower()).sum())

# State formatting
print("States not uppercase:",
      (df_sellers["seller_state"] != df_sellers["seller_state"].str.upper()).sum())

# Duplicate seller IDs
print("Duplicate seller IDs:",
      df_sellers["seller_id"].duplicated().sum())
```

```python
# Create cleaned copy
df_sellers_clean = df_sellers.copy()

# Standardize text data types
df_sellers_clean["seller_city"] = df_sellers_clean["seller_city"].astype("string")
df_sellers_clean["seller_state"] = df_sellers_clean["seller_state"].astype("string")

# Export
df_sellers_clean.to_csv(
    "../data/processed/sellers_clean.csv",
    index=False
)

print("Cleaned dataset exported successfully!")
print("Final Shape:", df_sellers_clean.shape)
```
