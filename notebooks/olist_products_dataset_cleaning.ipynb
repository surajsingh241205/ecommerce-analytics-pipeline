# Initial Data Exploration

```python
import pandas as pd
import numpy as np

df_products = pd.read_csv(
    "../data/raw/olist_products_dataset.csv",
    dtype={"product_id": "string"}
)

df_products.head()
```

```python
print("Shape:", df_products.shape)

print("\nColumn Information:")
df_products.info()

print("\nMissing Values:")
display(df_products.isnull().sum())

print("\nDuplicate Rows:", df_products.duplicated().sum())

print("\nStatistical Summary:")
display(df_products.describe(include="all"))
```

# Product category distribution

```python
display(
    df_products["product_category_name"]
    .value_counts()
    .head(15)
)
```

# Product category percentages

```python
display(
    df_products["product_category_name"]
    .value_counts(normalize=True)
    .head(15)
    .mul(100)
    .round(2)
)
```

# Product dimensions and weight

```python
display(
    df_products[
        [
            "product_weight_g",
            "product_length_cm",
            "product_height_cm",
            "product_width_cm"
        ]
    ].describe()
)
```

# Product photos distribution

```python
display(df_products["product_photos_qty"].value_counts().sort_index())
```

# Zero and invalid values

```python
numeric_cols = [
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

for col in numeric_cols:
    print(f"{col}:")
    print("Zero values:", (df_products[col] == 0).sum())
    print("Negative values:", (df_products[col] < 0).sum())
    print()
```

# Check missing records

```python
display(
    df_products[df_products.isnull().any(axis=1)]
)
```

```python
df_products_clean = df_products.copy()

# Convert category to string-friendly nullable type
df_products_clean["product_category_name"] = (
    df_products_clean["product_category_name"].astype("string")
)

# Check zero-weight products
zero_weight_products = df_products_clean[
    df_products_clean["product_weight_g"] == 0
]

display(zero_weight_products)

# Check records with missing dimensions/weight
missing_dimension_records = df_products_clean[
    df_products_clean[
        [
            "product_weight_g",
            "product_length_cm",
            "product_height_cm",
            "product_width_cm"
        ]
    ].isnull().any(axis=1)
]

display(missing_dimension_records)

# Final validation
print("Duplicate rows:", df_products_clean.duplicated().sum())
print("Duplicate product IDs:", df_products_clean["product_id"].duplicated().sum())
print("Missing values:")
display(df_products_clean.isnull().sum())
```

```python
df_products_clean.to_csv(
    "../data/processed/products_clean.csv",
    index=False
)

print("Products dataset cleaned and exported!")
print("Final shape:", df_products_clean.shape)
```
